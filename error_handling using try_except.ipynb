```python
# Keyword Method with iterrows()
# {new_key:new_value for (index, row) in df.iterrows()}

#TODO 1. Create a dictionary in this format:
{"A": "Alfa", "B": "Bravo"}

#TODO 2. Create a list of the phonetic code words from a word that the user inputs.

#creating a df from the csv file
df = pd.read_csv('nato_phonetic_alphabet.csv')
df.head()

#creating a dictionary from the df using iterrows and dictionary comprehension
my_dict = {row.letter:row.code for(index,row) in df.iterrows()}
my_dict

while True: #this is to run the a while loop indefinetly.
#usually you would have a condition here. But if you don't then while True is used to run an indeinite loop
#when you use while True,conditions to break the loop will be placed inside the lop using break and continue statements
    try:
        #getting input from the user
        my_input = input('What is your name?\n')

        #converting it into a list and all upper case to compare with keys in my_dict
        my_input = list(my_input.upper())
        my_input

        #getting the phonetic 
        phonetic_list = [my_dict[letter] for letter in my_input]
        print(phonetic_list)
        break #this is exit the while loop

    except KeyError:
        print('Sorry, please only use alphabets')
        continue #this will restart the loop, prompting the user for a new inupt. 
        #this way, you are forcing the user to only use alphabets
    
```
