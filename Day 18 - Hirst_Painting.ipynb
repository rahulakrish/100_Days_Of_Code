```python
%%writefile hirst_painting.py

from turtle import Turtle, Screen
import random

#create objects
t=Turtle()
screen = Screen()

#setting the colormode to interpret RGB colors
screen.colormode(255)

#function to generate random colors
def random_colors():
    r = random.randint(0,255)
    g = random.randint(0,255)
    b = random.randint(0,255)
    
    return (r,g,b)

#function to start drawing
def move_turtle(x,y): 
    #initializing y value
    y = 0
    
    #outer loop to go up rows
    for _ in range(0,5):
        
        #loop to draw each line
        for _ in range(0,5):
            t.dot(25, random_colors())
            t.penup()
            t.forward(100)
            t.pendown()
            t.dot(25, random_colors())
        
        #increment position
        t.penup()
        t.home()
        y += 50
        t.sety(y)        

screen.onscreenclick(move_turtle)
screen.mainloop()
```
