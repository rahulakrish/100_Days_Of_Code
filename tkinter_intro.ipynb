```python
from tkinter import *
```

```python
#create a window
#.Tk() creates the main window. In this window, you will created buttos, widgets, frames etc.
window = Tk()

#set the title
window.title("My First GUI Program")

#set the size
window.minsize(width=500, height=300)

#create a label
my_label = Label(text='My first Label')

#if you just run the above line, it will not show up.
#after creating the label, you then have to specify where in the window will it be laid out
#you do that using pack
my_label.pack() #default position is top


#packer is one of tkinter's geometry management mechanisims. 
#is used to manage the position of all components in the window

#if you just create a window and run, nothing will happen.
#to keep it open and listen for events, you should use .mainloop()
#this should always be the last line in your program. If you put this in between, nothing after it will run
window.mainloop()
```

Just like in `EXCEL VBA`, you can create `widgets` like `ComboBox, TextBox, Labels, Lists` etc.

A `widget` is a generic term for any of the aforementioned elements you want to add. You will come across this term a lot in the documentation.

Some terms to remember:

- `Window` – It is more like a container that holds other GUI elements in it.
- `Frame` – A  way to group related widgets in a rectangular area and add padding between widgets.
- `Label` – displays text on the screen.
- `Button` – A button can perform an action when clicked.
- `Entry` – A text box to take single-line text input from user.
- `Text` – A text box that allows multiline text entry.

##### Creating Widgets

`from tkinter import *`

This means you have imported all the classes including widgets from the `tkinter` module. 

If you're only creating a GUI with a few widgets i.e. if you just have a `Button and TextBox`, you can also choose to import only `Label and Entry` classes. 

```python
#creating a new window and configuration
window = Tk()
window.title("Widget Examples")
window.minsize(width=500, height=500)


#LABEL
label = Label(text="My first Label")
label.pack()


#BUTTON
def do_when_clicked():
    print('Button clicked')
    
button = Button(text="My first Button",command=do_when_clicked)
#command argument invokes the actions that you want performed when the 
#button is clicked
button.pack()

#ENTRIES
#setting the width of the textbox
entry = Entry(width=30)

#string argument will display the text in the box like a deault value
#END returns the index for where the string should start.
#END - the position just after the last character
#entry.insert(index, string)
entry.insert(END,string="Some text to begin with")
#.get() returns what's in the entry box. In this case, the string value.
print(entry.get())
entry.pack()


#TEXT
#used for hadnling multiple lines of text. is pretty much a large text editor
#in your window
text = Text(height=5, width=30,background='yellow')
text.focus()
text.insert(END, 'Example of multi-line text entry')
text.get('1.0',END)
text.pack()

#Spinbox
def spinbox_used():
    #gets the current value in spinbox.
    print(spinbox.get())
spinbox = Spinbox(from_=0, to=10, width=5, command=spinbox_used)
spinbox.pack()


#Scale
#Called with current scale value.
def scale_used(value):
    print(value)
scale = Scale(from_=0, to=100, command=scale_used)
scale.pack()

#Checkbutton
def checkbutton_used():
    #Prints 1 if On button checked, otherwise 0.
    print(checked_state.get())
#variable to hold on to checked state, 0 is off, 1 is on.
checked_state = IntVar()
checkbutton = Checkbutton(text="Is On?", variable=checked_state, command=checkbutton_used)
checked_state.get()
checkbutton.pack()

#Radiobutton
def radio_used():
    print(radio_state.get())
#Variable to hold on to which radio button value is checked.
radio_state = IntVar()
radiobutton1 = Radiobutton(text="Option1", value=1, variable=radio_state, command=radio_used)
radiobutton2 = Radiobutton(text="Option2", value=2, variable=radio_state, command=radio_used)
radiobutton1.pack()
radiobutton2.pack()


#Listbox
def listbox_used(event):
    # Gets current selection from listbox
    print(listbox.get(listbox.curselection()))

listbox = Listbox(height=4)
fruits = ["Apple", "Pear", "Orange", "Banana"]
for item in fruits:
    listbox.insert(fruits.index(item), item)
listbox.bind("<<ListboxSelect>>", listbox_used)
listbox.pack()
window.mainloop()

window.mainloop()
```
