- Selenium is a package that is used to automate web browser interactions.

- Used for automated testing of web applications but is also used for automated web scraping\
  and repetive web based tasks like taking out data from a spreadsheet and filling out a web form.

- Anyhting that a user would do i.e. type,click etc. can be automated using Selenium

- Like building a robot and telling it what to do on a browser.

- Is compatible with all different browsers like Chrome, Safari, Edge etc.

```python
#here, you will use NAME to find an element

#create an instance of the webdriver
driver = webdriver.Chrome()

#open the website
driver.get('https://www.python.org/')

#using NAME to find the search box on the webpage
search_box = driver.find_element(By.NAME, 'q')

#this will return a selenium element
print(search_box)

print('\n')

#use get_attribute method to if you want a particular attribute of an object
print(search_box.get_attribute('placeholder'))


#close the browser
driver.quit()



```

```python
#here, you will use ID to find an element. for eg: the GO box next to the search bar

#create an instance of the webdriver
driver = webdriver.Chrome()

#open the website
driver.get('https://www.python.org/')

#use ID for the GO box 
box = driver.find_element(By.ID, 'submit')


#use get_attribute to find it's name
print(box.get_attribute('name'))

print('\n')

print(box.size)


#close the browser after you're done
driver.quit()
```

```python
#here, you will use CSS_SELECTOR to find the link docs.python.org

#create an instance of the webdriver
driver = webdriver.Chrome()

#open the website
driver.get('https://www.python.org/')

#link lives under the class called documentation-widget

documentation_link = driver.find_element(By.CSS_SELECTOR,'.documentation-widget a')
print(documentation_link.text)

driver.quit()
```

```python
#here, you will use X_PATH to find the link docs.python.org
#when something is buried under a lot of elements and is hard to get to
#use X_PATH. works like naviagting files inside folders

#create an instance of the webdriver
driver = webdriver.Chrome()

#open the website
driver.get('https://www.python.org/')

#right-click on the element to get the X_PATH
link = driver.find_element(By.XPATH,'/html/body/div/footer/div[2]/div/ul/li[3]/a')
print(link.text)

driver.quit()
```

```python
#create a dictionary of all the events in the webpage

driver = webdriver.Chrome()

driver.get('https://www.python.org/')


#the event dates are under a class called event widget. 
#in the event widget, you get the time element
event_dates = driver.find_elements(By.CSS_SELECTOR, '.event-widget time')


#similarly, the names are in an anchor tag which is in a list
event_names  = driver.find_elements(By.CSS_SELECTOR, '.event-widget li a')

#create empty lists to later zip into a dict
dates = []
names = []

#loop to add the dates
for d in event_dates:
    dates.append(d.text)


#loop to add the names
for n in event_names:
    names.append(n.text)


#create the dict
my_dict = dict(zip(dates, names))

print(my_dict)

#close the browser
driver.quit()
```
