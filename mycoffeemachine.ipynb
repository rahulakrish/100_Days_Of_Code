```python
#dictionary for menu and ingredients

menu = {
        "espresso": {
            "ingredients": {
                "water": 50,
                "coffee": 18,
            },
            "cost": 1.5,
        },
        "latte": {
            "ingredients": {
                "water": 200,
                "milk": 150,
                "coffee": 24,
            },
            "cost": 2.5,
        },
        "cappuccino": {
            "ingredients": {
                "water": 250,
                "milk": 100,
                "coffee": 24,
            },
            "cost": 3.0,
        }
    }

#dictinary for ingredients quantity in the machine

resources = {
        "water": 300,
        "milk": 200,
        "coffee": 50,
        "money": 100
    }


#dictionary for calculating total cost
coin_values = {
        "penny": 0.01,
        "nickel": 0.05,
        "dime": 0.10,
        "quarter": 0.25,
    }
```

```python
#rebuilt the dictonary from above into a sperate Menu class for easier use

class Menu:
    def __init__(self):
    
        self.espresso = {"ingredients": {"water": 50, "coffee": 18},
                         "cost": 1.5}
        
        self.latte = { "ingredients":{"water": 200, "milk": 150, "coffee": 24},
                       "cost": 2.5}
        
        self.cappuccino =  {"ingredients": {"water": 250, "milk": 100, "coffee": 24},
                            "cost": 3.0
        }
    
        
```

```python
class CoffeeMachine:
    def __init__(self):
        
        self.resources = {
                            "water": 300,
                            "milk": 200,
                            "coffee": 50,
                            "money": 100
                         }
    #check resources to make the drink    
    def check_resources(self,user_input):
        
        #the user_input from the main loop is a string. To conver that into an attribute for Python to use,
        #you need this step. This will tell Python to use the string as an attribute in the class.
        drink = getattr(menu, user_input)
    
        #looping through the ingredients for the drink
        for ingredient,amount in drink['ingredients'].items():
            
            #if there is not enough, the can_make flag is set to False and used in the main loop
            #to print the missing ingredient
            if amount > self.resources[ingredient]:
                can_make = False
                missing_ingredient = ingredient
            
            #otherwise, can_make is set to True. missing_ingredient is set to an empty string since in the 
            #above condition, it is used to tell the user what's missing
            else:
                can_make = True
                missing_ingredient = ""               
            
            #return both values to use in the main loop
            return can_make,missing_ingredient
    
    #to check payment
    def process_money(self,user_input,quarter,dime,nickel,penny):
                 
            coin_values = {
                            "penny": 0.01,
                           "nickel": 0.05,
                           "dime": 0.10,
                           "quarter": 0.25
                          }
            
            total = (quarter*coin_values['quarter'])+(dime*coin_values['dime'])+(nickel*coin_values['nickel'])+(penny*coin_values['penny'])
           
        #retunring the value to use in the main loop
            return total
    
    #if a drink is dispensed,then the resources need to deducted by the relevant amounts
    def deduct_resources(self,user_input):
        
        #again, converting the user_input string into an attribute in the class
        drink = getattr(menu, user_input)
        
        for ingredient,amount in drink['ingredients'].items():
            self.resources[ingredient] -= amount
    
    #adding money to machine
    def add_money(self,user_input):
        drink = getattr(menu, user_input)
        self.resources['money'] += drink['cost']
```

```python
menu = Menu()
coffee_machine = CoffeeMachine()
```

```python
machine_on = True

#function to prompt user to pay. this will feed into the process_money method in the CoffeeMachine class.
#better to keep the functionality in the main loop than in the class. Easy to understand and makes for cleaner code.
def get_money():
    quarter = int(input('How many quarters?\n'))
    dime = int(input('How many dimes?\n'))
    nickel = int(input('How many nickels?\n'))
    penny = int(input('How many pennies?\n'))
    
    #the outputs will feed into the process_money method in the CoffeeMachine class
    return quarter,dime,nickel,penny


while machine_on:    
    user_input = input('What would you like to drink?\nEspresso\Latte\Cappuccino\n').lower()
    
    if user_input == 'off':
        machine_on = False
        
    elif user_input == 'report':
        print(coffee_machine.resources)
        machine_on = False
        
    else:
        #defining drink again here because you use it to check payment_success
        drink = getattr(menu, user_input)
        
        #getting the results from the check_resources method of the CoffeeMachine class
        can_make,missing_ingredient = coffee_machine.check_resources(user_input)
        
        #prompting for payment if all resources are available
        if can_make:
            quarter,dime,nickel,penny = get_money()
            
            #this calculates total money received
            payment = coffee_machine.process_money(user_input,quarter,dime,nickel,penny)
            
            #loop to compare received money to the cost
            if payment == drink['cost']:
                payment_success = True
                print('Here you go! Enjoy your drink!\n')
            
            #return change
            elif payment > drink['cost']:
                change = round((payment - drink['cost']),2)
                payment_success = True
                print(f"Here's your change: ${change}. Enjoy your drink!\n")
            
            #if full amount not received
            else: 
                print(f"Please pay the full amount. Here's your refund: ${payment}\n")
            
            #if resources are availble and money is paid, then the drink is dipsensed.
            #relevant resources deducted and money added to the machine
            if payment_success:
                      coffee_machine.deduct_resources(user_input)
                      coffee_machine.add_money(user_input)
         
        #if an ingredient is missing, then everything in-between is skipped and the loop ends.    
        else:
            print(f'Sorry, Not enough {missing_ingredient}')
            machine_on = False
```
