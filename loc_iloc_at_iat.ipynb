## `loc and iloc`

- loc - label based indexing
- iloc - integer based indexing


```python
import pandas as pd
df = pd.read_csv('2018_Central_Park_Squirrel_Census_-_Squirrel_Data.csv')
df.head()
```

```python
df.info()
```

### `.loc` syntax: `df.loc[row_label,column_label]`

`.loc` is known as label-based indexing. 

You use it only when you want to use the name of the rows or columns in the `df`.

`row_label` refers to the index of the dataframe.

Therefore, if you want to get a specific row, you will refer to the index. In this case, the `df` has a numeric value

Here, For `row[0]`, which is the first row in the `df`, you will get all the column values:

```python
df.loc[0]
```

`df.loc[0]` will return a `Series` data type value

If you want mulitple rows, then:

```python
df.loc[0:2]
```

In this case, `df.loc[0:2]` will return a `DataFrame`

Sometimes you can have `string indexes` in the `DataFrame`:

```python
df = pd.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["NY", "LA", "SF"]
}, index=["a", "b", "c"])

df
```

Here the `.loc` syntax will look like:

```python
df.loc['a']
```

For string indexes, when you want to multiple rows, you have to use `,` to seperate the first and the last row you specify.

`:` is only for numerical indexes.

And, you have to enclose the rows in double brackets:

```python
df.loc[['a','b']]
```

### Using `.loc` for rows and columns 

```python
#a specific row and column
df.loc[0,'Hectare']
```

```python
#multiple rows and a column
df.loc[0:2,'Hectare']
```

```python
#multiple rows and multiple columns
df.loc[0:2,['Age', 'Hectare']]
```

### `.loc` syntax quirk

If you want to a specific column using '`.loc`, then you cannot leave the row argument blank. 

You always have to specify which rows you want for the specific column. 

If you want all the rows, then use `:`:

```python
df.loc[:,'Age']
```

If you simply type: `df.loc['Age']`, python will assume the `Age` to be row label. You have to specify something in the row argument.

But it doesn't work that way for `rows`. When you want a particular row, you can leave the column argument blank. Pyhton will just return all the columns for that specific row, unless you specify which columns you want. 

This is an important quirk to remember.

### `iloc` syntax: `df.iloc[row_index(es), column_index(es)]

`iloc` refers to integer based indexing.

Here, you do not use the names of the rows and columns.

Instead, you refer to their positions in the `DataFrame`, i.e. 1st row, 5th column etc.

```python
#single row
df.iloc[0]
```

```python
#multiple rows
df.iloc[0:2]
```

```python
#multiple rows and a single column
df.iloc[0:2,2]
```

```python
#multiple rows and columns
df.iloc[0:2, 3:5]
```

Just like in `.loc`, `.iloc` has the same quirk i.e. you cannot leave the row argument blank, when you are looking for a specific column. You need to specifiy something in the row argument.

### If you want just one particular value from the DataFrame, then you can use `.at` or `.iat`

```python
df.at[0,'Shift']
```

This is will return a `str` data type value, which you can use further down the line if you need to.

This is similar to:

```python
df.loc[0,'Shift']
```

The difference between `.at` and `.iat` is similar to `.loc` and `.iloc` i.e. label based indexing and integer based indexing
