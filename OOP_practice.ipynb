✅ 1. Create a Book Class

Attributes: title, author, pages, current_page
Methods:

read(pages) → increases current_page

restart() → sets current_page to 1

Goal: Understand updating internal state.

```python
class Book():
    #setting up attributes i.e what it has...
    def __init__(self,title,author,pages,current_page):
        self.title = title
        self.author = author
        self.pages = pages
        self.current_page = current_page
    
    #setting up methods i.e. what it can do
    #here, it can dispaly the current page number
    def read(self,pages):
        self.current_page = self.current_page + pages
        print(f'New current page is {self.current_page}')
    
    #here, it can reset the page number to 0.
    def restart(self):
              self.current_page = 0
              print(f'Current page reset to {self.current_page}')
```

```python
#create an instance of the book class
new_book = Book("Hello","Me",200,50)
new_book
```

```python
#check out an attribute
new_book.title
```

```python
#employ a method
new_book.read(50)
```

```python
#employ another method
new_book.restart()
```

✅ 2. Build a BankAccount Class

Attributes: owner, balance
Methods:

deposit(amount)

withdraw(amount)

show_balance()

Goal: Practice modifying and checking attribute values.

```python
class BankAccount():
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
        print(f'New account created. Account owner is {self.owner}. Account balance is {self.balance}')
    
    def deposit(self,amount):
        self.balance += amount
        print(f'New balance is ${self.balance}')
        
    def withdraw(self,amount):
        if amount > self.balance:
            print('Sorry. Insufficient funds')
        else:
            self.balance -= amount
            print(f'New balance is ${self.balance}')
            
    def show_balance(self):
        print(f'Current balance is ${self.balance}')    
```

```python
#create an instance
new_account = BankAccount("XYZ", 1000)
```

```python
#method 1 - make a deposit
new_account.deposit(500)
```

```python
#method 2 - withdraw money
new_account.withdraw(2000)
```

```python
#method 2 - withdraw money
new_account.withdraw(100)
```

```python
#method 3 - check balance
new_account.show_balance()
```

✅ 3. Make a Student Class

Attributes: name, grades (list)
Methods:

add_grade(grade)

average() → return the mean grade

Goal: Working with lists stored inside an object.

```python
class Student():
    def __init__(self,name,grades):
        self.name=name
        self.grades = grades
        
    def average(self):
        print(f'Average grade for {self.name} is {sum(self.grades)/len(self.grades)}')
    
```

```python
abc = Student('abc',[30,40,50,50])

#access the attributes
print(abc.name)
print(abc.grades)

#access the methods
abc.average()
```

```python
#modify the grades attribute
abc.grades.append(85)
print(abc.grades)

abc.average()
```

✅ 4. A Simple Pet Class

Attributes: name, energy
Methods:

play() → decreases energy

feed() → increases energy

status() → returns "happy", "tired" etc. based on energy

Goal: Use attributes to influence behavior.

```python
class Pet():
    def __init__(self,name,energy):
        self.name = name
        self.energy = energy
    
    def play(self,amount):
        self.energy -= amount
        
    def status(self):
        if self.energy <= 20:
            print("I'm too tired!")
        elif self.energy > 20 and self.energy <= 100:
            print("Let's play! 😊")
            
    def feed(self,amount):
        if amount + self.energy > 100:
            print("That's too much food!")
        elif amount + self.energy < 100:
            print("Not enough, but ok!")
        elif amount + self.energy == 100:
            print ("Yay! Let's go!")
    
```

```python
toffee = Pet('toffee',45)

print(toffee.name)
print(toffee.energy)
```

```python
toffee.play(25)
toffee.status()
```

```python
toffee.play(10)
toffee.status()
```

```python
toffee.feed(100)
```
