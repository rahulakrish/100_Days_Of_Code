```python
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from time import time
```

```python
driver = webdriver.Edge()

#this website bypasses the cookies pop up
driver.get('https://ozh.github.io/cookieclicker/')

#to click on the english language
#this is the selenium way to do it.
#you can also just put sleep(2) to pause the program after the website loads to wait for the element to become active
#the selenium way translates to: “Wait up to 2 seconds for the English language button to become clickable, then return that element.”
#if the element becomes visible in under 2 seconds, then it click on it. Sleep(2), will pause the program for 2 seconds regardless of the element visibility.
#hence, the selenium way is faster
english_element =  WebDriverWait(driver,2).until(EC.element_to_be_clickable((By.ID,'langSelect-EN')))
english_element.click()

#find the big cookie
cookie_button = driver.find_element(By.ID,'bigCookie')

#set how long you want the program to run
duration =  60

#define start time
start_time = time()

#set interval of 5 seconds for the upgrade check
next_upgrade_check = time() + 5

#just to see how many iterations of the loop run
iterations = 0

#get the loop running
while time() - start_time < duration:
    cookie_button.click()
     
    #find the element to get the number of cookies
    cookie_count_element = driver.find_element(By.ID,'cookies')
    cookie_count = int(cookie_count_element.text.split()[0])
    
    #check for upgardes
    if time() >= next_upgrade_check:
        #use find_elements(s) to get all the upgrades
        products = driver.find_elements(By.CSS_SELECTOR,'.product.unlocked.enabled')
        
        #click on the last one because that will be the most expensive and give you the most as well
        if products:
            products[-1].click()

        next_upgrade_check = time() + 5

    iterations +=1

#just to check how long the loop ran
end_time  = time()

print(end_time - start_time)
print(f"number of iterations:{iterations}")
```
