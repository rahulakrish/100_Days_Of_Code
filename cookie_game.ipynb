```python
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

from time import sleep,time
```

```python
driver = webdriver.Chrome()

#this website bypasses the cookies pop up
driver.get('https://ozh.github.io/cookieclicker/')

count = 0

#doing this to hold the program from running to give the webpage time to load
sleep(5)

#to click on the english language
eng =  driver.find_element(By.XPATH,'//*[@id="langSelect-EN"]')

eng.click()

sleep(1)


while count <= 20:

    #to click on the big cookie
    cookie_button = driver.find_element(By.ID, 'bigCookie')

    cookie_button.click()

    #holding the program here for the  webpage to update the element that counts the cookies
    sleep(0.5)
    
    #getting the number of cookies
    cookie_count_element = driver.find_element(By.ID, 'cookies')

    number_of_cookies = cookie_count_element.text 

    count += 1

    # print(number_of_cookies)

#get the actual number from the string

number_of_cookies = number_of_cookies.split()
number_of_cookies = int(number_of_cookies[0])
print(number_of_cookies)

sleep(0.5)

product = driver.find_element(By.CSS_SELECTOR, '.product.unlocked.enabled' )

print(product.text)



```
