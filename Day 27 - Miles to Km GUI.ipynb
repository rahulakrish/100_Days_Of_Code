```python
from tkinter import *
```

```python
#create a window
window = Tk()

window.minsize(width=500,height=500)
window.title("Mile to KM Converter")

#function to convert miles to km when the button is clicked.
#will be passed as an argument when defining the button.
def button_clicked():
    miles = float(entry_miles.get())
    km = round(miles * 1.6,2)
    entry_km.delete(0, END)
    entry_km.insert(0, km)

#creating the entry box for the user to input the number of miles.
    
#padx and pady are used to add some extra spacing around each widget.
#once you get the position right, then add some spacing to imporve the layout. 
#don't look to re-position the widget. use padding instead.
entry_miles = Entry(width=10)
entry_miles.grid(row=2, column=4,padx=20,pady=10)


#label for the miles entry box
miles_label=Label(text='Miles')
miles_label.grid(row=2,column=5)

#label for the km entry box
text_label = Label(text='is equal to')
text_label.grid(row=5,column=0,padx=10,pady=10)

#entry box for the output in kms to be displayed
entry_km = Entry(width=10)
entry_km.grid(row=7,column=4,padx=20,pady=10)

#label for the km entry box
km_label = Label(text='Km')
km_label.grid(row=7,column=5)

#defining the buttom
button = Button(text='Calculate', command=button_clicked)
button.grid(row=10,column=5,pady=20)

window.mainloop()
```

One important thing about the `grid` manager:

It is always relative to another widget. Hence, if you specify `row=3, column=5` for the first widget without any other widget in the window, the first widget will always show up at `(0,0)`.

Hence, the best way to start is always at `(0,0)` for the first widget, build the other widgets and then start moving them into position.
