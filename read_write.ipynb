### Read-Write Practice

This is a way yo interact with other files using Python i.e. open, read and write.

It's not enough if you just open the file. Once you open the file, you then have to specify what you want to do i.e. read, write or append(add new text)

This is the syntax:

`with open(file_name, mode)`

There are more args , but for now this is all you need. Mode is where you specify what you want to do:
- ' r ' - read
- ' w ' - write
- ' a ' - append


```python
#write something to a file
with open("new_file.txt",mode='w') as file:
    file.write("\nHello")
```

```python
#read it's contents
with open("new_file.txt", mode = 'r') as file:
    contents = file.read()
    print(contents)
```

`files.read` will return the contents as a single string.

`files.readlines()` will return the contents as a list. 

If you have many lines, each line will be an item in the list

```python
#add more text to the file
with open("new_file.txt", mode = 'a') as file:
    file.write('\nNew entry')
```

```python
#read the updated file
with open("new_file.txt", mode = 'r') as file:
    contents = file.read()
    print(contents)
```

Even if the file does not exist, python will create the file and write to it:

```python
#data_file.txt does not exist currently
#Notice, here in the with statement itself, you are specifying the name and the type of the file i.e data.txt
#the next line is what goes in the file
with open('data_file.txt',mode='w') as file:
    file.write('Does this work?')
```

```python
with open ('data_file.txt') as file:
    print(file.read())
```

But it will not work for read mode. For read mode, the file should exist.

```python
with open('text.txt',mode='r') as file:
    contents = file.read()
```

### `with`

`with` keyword is used to help automatically close the file once the operation is complete. Otherwise, it will be left open until the user closes it. 

You can also open and close files without using `with`. But remeber to close it.

```python
#opening the file and reading
file = open('new_file.txt',mode='r')
print(file.read())
```

```python
#closing the file
file.close()
```

### Accessing a file that's in a different location from your current working directory

### Absolute path

Your current working directory is as follows:
`C:\Users\e40050338\Documents\Python`

To access a file that's for eg. in the Desktop, following is the syntax:

```python
with open('/Users/e40050338/Desktop/new_file.txt') as file:
    print(file.read())
```

You can get the path of the file by right-clicking on the file and selecting `Properties`. Replace the backslashes with forward-slash. The first forward-slash in this signifies the root directory i.e. in this case `C:`

This method uses the `Absolute Path`. You are accessing the file from the root directory using it's full path. While this is easier to follow, it's becomes tedious if you have many files and you have to manually move them around.

### Relative Path

In such a situation, it's better to use the `Relative Path` method.

For relative path, if you want to access a file that's in your current working directory, this is the following syntax:

`./filename`.

The `.` signifies look in the current dictionary for this file.

***

If you are looking for file that's in a sub-folder in your current working directory, i.e. you're **moving down the directory tree**, this is the following syntax:

`./sub-folder_name/file_name`

***

Now, if you want to **move up the directory tree**, this is the following syntax:

`../file_name`

`..` is telling Python to move up a folder and get to the file

To go up one folder, it is `../file_name`

To go up two folder, it is `../../file_name`

or

`../../sub_folder_name/file_name`. Here, you are going up two folders and looking for the file in the sub-folder that's in the second folder. 🙃
