```python
from tkinter import *
import requests
```

```python
#constant set-up
BACKGROUND_COLOR = "#B1DDC6"
background_image_path = r"C:\Users\e40050338\Documents\Python\Day 32 - API End Ppoints + ISS Notifier\background.png"
kanye_image_path = r"C:\Users\e40050338\Documents\Python\Day 32 - API End Ppoints + ISS Notifier\kanye.png"
```

```python
#function to make api calls and get quotes
def get_quote():
    #using the requests library to make calls
    response = requests.get(url='https://api.kanye.rest')
    #converting the data into json format
    data = response.json()
    #modifying the quote in the canvas widget
    canvas.itemconfig(quote,text=data['quote'])
    
#-------------------------------------------UI Setup----------------------------------------------#

#create a window
window = Tk()
window.title('Kanye Quote for the Day')
window.config(padx=50, pady=10)

#create the canvas widget
canvas = Canvas(width=500, height=500, highlightthickness=0)

#use PhotoImage class to import images onto the Canvas
background_image = PhotoImage(file=background_image_path)
kanye_image = PhotoImage(file=kanye_image_path)

#create the image
canvas.create_image(250,250,image=background_image)

#create the quote 
quote = canvas.create_text(250,250,text="Kanye quote goes HERE", width=250, font=("Arial", 25, "bold"))

#position the canvas using grid
canvas.grid(row=1,column=1)

#create a button and invoke the get_quote function to get new quotes
get_quote_button = Button(image=kanye_image,command=get_quote,highlightthickness=0)

#positiont the button using grid
get_quote_button.grid(row=2,column=1)

window.mainloop()
```
