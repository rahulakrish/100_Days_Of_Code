
Attributes are what the object has.
Attributes are defined this way:  **def__init__(self,*)**

The stars is a parameter that you can pass into the attribute if you need to.

Methods are what the object can do.

Methods are created the same way you create a function i.e. using **def(self)**


__self__  refers to the object that will be created when you initialize the class.
Whenever you are defining attributes or methods in a class, you always have to refer to the object that will be created using 
the __self__ keyword.

Even when you don't have parameters, Python will still pass the object into the attribute or method.If you do not specify __self__ and just leave it blank, you will get an error saying 1 argumnet was passed but 0 was specified.

```python
question_data = [
{"text": "A slug's blood is green.", "answer": "True"},
{"text": "The loudest animal is the African Elephant.", "answer": "False"},
{"text": "Approximately one quarter of human bones are in the feet.", "answer": "True"},
{"text": "The total surface area of a human lungs is the size of a football pitch.", "answer": "True"},
{"text": "In West Virginia, USA, if you accidentally hit an animal with your car, you are free to take it home to eat.", "answer": "True"},
{"text": "In London, UK, if you happen to die in the House of Parliament, you are entitled to a state funeral.", "answer": "False"},
{"text": "It is illegal to pee in the Ocean in Portugal.", "answer": "True"},
{"text": "You can lead a cow down stairs but not up stairs.", "answer": "False"},
{"text": "Google was originally called 'Backrub'.", "answer": "True"},
{"text": "Buzz Aldrin's mother's maiden name was 'Moon'.", "answer": "True"},
{"text": "No piece of square dry paper can be folded in half more than 7 times.", "answer": "False"},
{"text": "A few ounces of chocolate can to kill a small dog.", "answer": "True"}
]
```

```python
#building the Class for Question
#takes two attributes text and answer

class Question:
    def __init__(self,text,answer):
        self.text = text
        self.answer = answer
```

```python
#Creating a list of Question objects from the data

#initialize an empty list
question_bank = []

#build a for loop to go thru each item in question_data and append it to the question_bank
#you will end up with a list of Objects created using the Question class with 2 attributes: text and answer

for question in question_data:
    question_text = question['text']
    question_answer = question['answer']
    question = Question(question_text,question_answer)
    question_bank.append(question)
    
question_bank[0]
```

```python
#you can access the attributes like so:
print(question_bank[0].text)
print(question_bank[0].answer)
```

```python
#creating the quiz_brain class
#Initilaize the question number to 0
#initialize the question_list to an output

class QuizBrain:
    #QuizBrain class will have two attributes: question_number and question_list.
    #q_list here is the parameter. You will pass the question_bank created above into the object as a parameter
    def __init__(self,q_list):
        self.question_number = 0
        self.question_list = q_list
        self.score = 0
        
    #this is a method. This is what the QuizBrain class can do. It can ask questions
    #You want to start at the first question obviously. You get the question number from the attribute you defined above. 
    def next_question(self):
        current_question = self.question_list[self.question_number]
        
        #incrementing the question number to ask the next question
        self.question_number += 1
        
        #asking the question
        user_answer = input(f"Q.{self.question_number}: {current_question.text}: True\False\n")
        
        #checking the answer
        self.check_answer(user_answer,current_question.answer)
    
    #this is a method to keep aksing questions. by using the return statement, you will either get a True or False value.
    #remeber, about "truthy" values in Python
    def still_has_questions(self):
        return self.question_number < len(self.question_list)
    
    def check_answer(self,user_answer,correct_answer):
        if user_answer.lower() == correct_answer.lower():
            print("That's correct!")
            self.score += 1
                  
        else:
            print("That's wrong!")
            print(f'The correct answer is: {correct_answer}.')
                  
        print(f'Your current score is: {self.score}/{self.question_number}')
        
        print('\n') 
```

```python
#creating an object.
#the question bank is the parameter that is an attribute of the object
qz = QuizBrain(question_bank)
```

```python
#loop to keep asking questions
#to access a method in a class, you have to call it like a function
#here the while loop will keep running till it runs out of questions to ask

while qz.still_has_questions():
    qz.next_question()
    
print("You've completed the quiz!")
print(f"Your final score is {qz.score}/{len(qz.question_list)}")
```
