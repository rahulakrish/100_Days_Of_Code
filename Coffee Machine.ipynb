## Coffee Machine

```python
#dictionary for menu and ingredients

menu = {
        "espresso": {
            "ingredients": {
                "water": 50,
                "coffee": 18,
            },
            "cost": 1.5,
        },
        "latte": {
            "ingredients": {
                "water": 200,
                "milk": 150,
                "coffee": 24,
            },
            "cost": 2.5,
        },
        "cappuccino": {
            "ingredients": {
                "water": 250,
                "milk": 100,
                "coffee": 24,
            },
            "cost": 3.0,
        }
    }

#dictinary for ingredients quantity in the machine

resources = {
        "water": 300,
        "milk": 200,
        "coffee": 50,
        "money": 100
    }


#dictionary for calculating total cost
coin_values = {
        "penny": 0.01,
        "nickel": 0.05,
        "dime": 0.10,
        "quarter": 0.25,
    }
```

```python
#function to check resources i.e. if the machine has enough qty of each ingredient to make the user's choice of drink
def resource_check(user_choice):
    
    #checking ingredients for espresso
    if user_choice == 'espresso':
        if resources['water'] < menu[user_choice]['ingredients']['water'] and \
           resources['coffee'] < menu[user_choice]['ingredients']['coffee']:
            
            print('\n')
            print('Sorry! Not enough Water and Coffee left!')
            resource_status = False
    
        elif resources['water'] < menu[user_choice]['ingredients']['water']:
            print('\n')
            print('Sorry! Not enough water')
            resource_status = False
        
        elif resources['coffee'] < menu[user_choice]['ingredients']['coffee']:
            print('\n')
            print('Sorry! Not enough Coffee')
            resource_status = False
            
        else:
            resource_status = True
#setting a resource_status flag to be used in the last loop. When the flag is False, that means the machine is out of an
# ingredient to make the drink. If it's True, then the machine will dispense the drink.
    
    #checking ingredients for latte
    elif user_choice == 'latte':
        if resources['water'] < menu[user_choice]['ingredients']['water'] and resources['coffee'] < menu[user_choice]['ingredients']['coffee'] and \
           resources['milk'] < menu[user_choice]['ingredients']['milk']:
                print('\n')
                print('Sorry! Not enough Water, Milk and Coffee left!')
                resource_status = False
                
        elif resources['water'] < menu[user_choice]['ingredients']['water']:
            print('\n')
            print('Sorry! Not enough water left!')
            resource_status = False
            
        elif resources['milk'] < menu[user_choice]['ingredients']['milk']:
            print('\n')
            print('Sorry! Not enough milk left!')
            resource_status = False
            
                    
        elif resources['coffee'] < menu[user_choice]['ingredients']['coffee']:
            print('\n')
            print('Sorry! Not enough coffee left!')
            resource_status = False
            
        else:
            resource_status = True

    #checking ingredients for cappuccino          
    elif user_choice == 'cappuccino':
        if resources['water'] < menu[user_choice]['ingredients']['water'] and \
           resources['coffee'] < menu[user_choice]['ingredients']['coffee'] and \
           resources['milk'] < menu[user_choice]['ingredients']['milk']:
                print('\n')
                print('Sorry! Not enough Water, Milk and Coffee left!')
                resource_status = False
                
        elif resources['water'] < menu[user_choice]['ingredients']['water']:
            print('\n')
            print('Sorry! Not enough water left!')
            resource_status = False
            
        elif resources['milk'] < menu[user_choice]['ingredients']['milk']:
            print('\n')
            print('Sorry! Not enough milk left!')
            resource_status = False
            
                    
        elif resources['coffee'] < menu[user_choice]['ingredients']['coffee']:
            print('\n')
            print('Sorry! Not enough coffee left!')
            resource_status = False
            
        else:
            resource_status = True
 #using the return statement to get resource_status outside the function
    return resource_status 

#function to check if correct amount is paid for the drink
def money_check(user_choice):
        
        print(f"Cost of {user_choice} is ${menu[user_choice]['cost']}")

        quarters = int(input('How many quarters?\n'))
        dimes = int(input('How many dimes?\n'))
        nickels = int(input('How many nickels?\n'))
        pennies = int(input('How many pennies?\n'))

        total = quarters*coin_values['quarter'] + dimes*coin_values['dime'] + nickels*coin_values['nickel'] + pennies*coin_values['penny']

        if total == menu[user_choice]['cost']:
            print(f"Thanks! Here's your {user_choice}! Enjoy!")
            print('\n')
            resources['money'] += total #adding the total to the machine resource
            money_status = True
            
        elif total < menu[user_choice]['cost']:
            print("Sorry! That's not enough money. Money refunded")#refund the money if user does not pay the right amount
            money_status = False
            
            
        elif total > menu[user_choice]['cost']:
            change = round((total - menu[user_choice]['cost']),1)
            print(f"Thanks! Here's your change: ${change}. Enjoy your drink!")#return change to the user
            print('\n')
            resources['money'] += menu[user_choice]['cost'] #adding the total to the machine resource
            money_status = True
#similar to resource_status, seetting a flag for money_status here. If it's False, then the loop stops. If not, then the machine
#will dispense the drink
        return money_status
# user return statement to get the value of money_status and total outside the function.
```

```python
#final loop that brings everyhting together

while True:
    #prompt user for drink option
    user_choice = input('What would you like to drink?\nEspresso/Latte/Cappuccino?\n').lower()
    
    #secret word to shut off the machine for maintenance
    if user_choice == 'off':
        
        print('\n')
        
        print('Goodbye')
        
        break
    #another secret word to check ingerdient qty currently in the machine    
    if user_choice == 'report':
        
        print('\n')
            
        print(f"Water: {resources['water']}ml")

        print(f"Milk: {resources['milk']}ml")

        print(f"Coffee: {resources['coffee']}g")
        
        print(f"Money: ${resources['money']}")
        
        continue
    #input valdation    
    if user_choice not in ['espresso', 'latte', 'cappuccino']:
        
        print('\n')
        
        print('Sorry! Not a valid option! Please try again.')
        
        continue
        
    # Step 1: Check resources first using the function. Here resource_status will either be True or False
    resource_status = resource_check(user_choice)
    
    # Step 2: Skip if not enough. No need to even check/ask for money. The loop ends here.
    if resource_status == False:
        break
    
    # Step 3: Ask for payment only using the function only if enough resource_status from the above is True. Here, like the 
    # resource_status, money_status will either be True or False
    money_status = money_check(user_choice)
    
    # Step 4: If only both statuses are true, then make the drink, deduct resources. Money is added to machine in the function itself
    if money_status == True:
        if user_choice == 'espresso':
            resources['water'] -= menu[user_choice]['ingredients']['water']
            resources['coffee'] -= menu[user_choice]['ingredients']['coffee'] 
        else:
            resources['water'] -= menu[user_choice]['ingredients']['water']
            resources['coffee'] -= menu[user_choice]['ingredients']['coffee']
            resources['milk'] -= menu[user_choice]['ingredients']['milk']
       
```
