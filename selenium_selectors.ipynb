```python
"""
Things become a lot more difficult when the website get complicated.

There are 2 ways to locate an element: look at the whole page (or) look inside an element. For eg: a particular link nested in a div tag.

Understanding how to locate nested HTML elements is a crucial milestone in mastering Selenium.
When elements don't have unique IDs or names, looking at them in isolation won't work.
Instead, you have to look at their family tree (their relationship to the elements around them).

In HTML, elements inside other elements are called children or descendants, while the outer elements are parents or ancestors.

<div id="login-form">
    <div class="input-group">
        <input type="text" name="username">
    </div>
</div>

If you just search for name="username", it might work, but if there are multiple username fields on the page (like a registration form hidden in the background),
Selenium might grab the wrong one. You need to tell Selenium: "Go inside the login form, then find the username field."

XPath is incredibly powerful for nested elements because it allows you to travel both down to children and up to parents.

To travel downwards using descendants, use a double slash //:

username_input = driver.find_element(By.XPATH,value="//div[@id='login-form']//input[@name='username']")

When you want to say: "Find a specific parent, then look anywhere inside it for a child."

The Formula: //[Parent]//[Child]

The Slashes Meaning: * // at the start means "search the whole page".

// in the middle means "search anywhere inside the parent".

<div class="card-body">
    <div>
        <button>Submit</button>
    </div>
</div>

The XPath: //div[@class='card-body']//button

Translation: Find the div with the class card-body, look inside it (skipping the middle div), and find the button.

go_button = driver.find_element(By.XPATH, "//a[@aria-label='start speed test - connection type multi']")

breakdown:

go_button =
This creates a variable named go_button. 
Think of it like a shortcut or a container. 
Once Selenium finds the button on the page, it stores a reference to it in this variable so you can use it later.

driver.find_element(...)
This tells your automated browser (driver) to look at the current webpage and search for a single element.

By.XPATH
This tells Selenium how to search. You are telling it: "Hey, I'm going to give you an XPath coordinate, use that blueprint to scan the HTML tree."

"//a[@aria-label='start speed test - connection type multi']"
This is the actual XPath string (the blueprint). Let's decode this specific code:

// means: Search anywhere on the entire webpage from top to bottom.

a means: Specifically look for an anchor tag (<a>), which is used for links or buttons.

[@... ] The square brackets mean: "Filter these results by a specific attribute."

aria-label='...' This targets the exact accessibility label we talked about earlier. Selenium looks for the <a> tag that has that exact text inside its aria-label attribute.


"""

;
```

```python
"""
CSS Selectors

| Strategy     | Symbol / Syntax | HTML Example              | CSS Selector   |
|--------------|-----------------|---------------------------|----------------|
| By ID        | #               | '<input id="search">'     | #search        |
| By Class     | .               | '<div class="card">'      | .card          |
| By Attribute | [attr='value']  | '<input type="email">'    | [type='email'] |
| By Tag Name  | Just the tag    | '<button>Submit</button>' | button         |

Advanced Attribute Matching

Starts With (^=): Matches the beginning of a value.

HTML: <div id="post_10482">

Selector: div[id^='post_']

Ends With ($=): Matches the end of a value.

HTML: <input name="user_email_address">

Selector: input[name$='_address']

Contains (*=): Matches a substring anywhere inside the value.

HTML: <button class="sc-axgKjh kYgUvD"> (Dynamic React classes)

Selector: button[class*='kYgUvD']

"""

;
```

```python
"""
Spaces in CSS selectors:

In a CSS selector, a single space represents a descendant relationship.

It acts as a structural path, telling the browser or Selenium: "Find the element on the left first, then look anywhere inside it to find the element on the right."

<div class="start-button"> 
    
    <div>
        <span class="start-text">GO</span> 
    </div>
    
</div>

driver.find_element(By.CSS_SELECTOR,".start-button .start-text")

.start-button (Before the space): The starting point. Find the outer container element with this class.

  (The Space): "Look inside the HTML tags nested within that container."

.start-text (After the space): The final target. Find an element with this class only if it lives inside the starting point.


"""
```

"""
Classes can have long names. For eg:
class="result-data-large number result-data-value download-speed"

The secret lies in understanding how browsers read HTML classes, and how the CSS dot (.) syntax actually works.

Spaces in HTML Mean "Multiple Separate Classes"

To human eyes, that long class string looks like one single, massive name.
But to a web browser, a space inside a class="" attribute acts as a comma.
The element in your screenshot doesn't have one long class name; it actually has four completely separate classes assigned to it at the same time:

result-data-large (Controls the font size)

number (Applies number-specific styling)

result-data-value (Aligns or pads the data layout)

download-speed (Identifies this specific metric)

When we use a dot in a CSS Selector (like .download-speed), we are telling Selenium: "Search through the list of classes on every element. If you find an element that has this specific class name in its list, grab it."

Selenium doesn't care about the order of the list, nor does it care how many other classes are sharing the space.
