Above code is how you would normally write it, but there are issues running `Tkinter` .
If you try to run the code a second time, you will get the following error:

**TclError: invalid command name ".!canvas"**

This is because the the `Tkinter` window is closed by `screen.exitonclick()`, but Python still has a reference to it.
Hence, it cannot draw on a canvas that's closed and hence the above error.

Now, you can create a new instance of the `Turtle` and `Screen` classes, but that also throws an error.
This is because the underlying `Tkinkter` interpreter that `Turtle` is built on does not clean itself up properly and hence
you will get errors when trying to run it again.

The `Turtle` module is not designed for persistent interactive environments like Jupyter. It works best in standalone scripts (.py files) run from a terminal.

#### Dashed Line

```python
%%writefile dashed_line.py
from turtle import Turtle, Screen

t = Turtle()
screen = Screen()


def move_turtle(x,y):
    for _ in range(4):
        t.forward(20)
        t.penup()
        t.forward(20)
        t.pendown()
        t.forward(20)


#oncreenclick method gives you more control.Turtle will move only after you click on the screen.
#it takes tow arguments: x and y. so you have to define it as a function with x and y parameters even if you don't use them.
#I suppose, when you click on the screen, the x,y coordinates of that point are paseed into the function and it starts
screen.onscreenclick(move_turtle)


#this is to keep the screen on even after execution. You have to manually close the screen.
screen.mainloop()
```

#### Multiple Shapes

```python
%%writefile shapes.py
from turtle import Turtle,Screen

t = Turtle()
screen = Screen()

#defining number of sides for each shape
number_of_sides = [3,4,5,6,7,8,9,10]

#empty list to store the angles the turtle has to turn
turn_signals = [] 

#angle = number of sides in each shape/360
#add it to the list above
for side in number_of_sides:
    turn_signals.append(360/side)

#list of colors. each shape will be in a different color
colors = ['red','dark green','blue','yellow','purple1','DarkOrange1','cyan','black']

def move_turtle(x,y):
    
    #setting the start points for the loop
    color_index = 0
    start_shape = 3

    #start off by facinf the right direction
    for signal in turn_signals:
        #set the number of sides for each shape
        for sides in range(0,start_shape):
            #set the color
            t.color(colors[color_index])
            #set direction
            t.right(signal)
            #start moving
            t.forward(100)
        #increase the number of sides by 1 for the next shape
        start_shape += 1
        #draw the next shape in a new color
        color_index += 1
    
screen.onscreenclick(move_turtle)
screen.mainloop()
```

```python
%%writefile move_random.py

import random
from turtle import Turtle,Screen

t = Turtle()
screen = Screen()


#to make each movement generate a random color.
#to use RGB values, you first have to set the colormode to 255, to tell the turtle object to interpret the values 
#b/w 0 to 255
screen.colormode(255)

#setting the shape,size and speed of the turtle object
t.shape('circle')
t.shapesize(0.5)
t.speed(2)


#setting distances for forward and backward and also right and left turn angles
distance = 50
angle = 90


#defining functions
def move_forward():
    t.forward(distance)
    
def move_back():
    t.backward(distance)
    
def turn_right():
    t.right(angle)
    
def turn_left():
    t.left(angle)
    
#function to make the turtle move
def move_turtle(x,y):
    moves = [move_forward, move_back, turn_right, turn_left]
    for _ in range(0,100):
        #generating random RGB values b/w 0 and 255
        r = random.randint(0,255) 
        g = random.randint(0,255)
        b = random.randint(0,255)
        
        #setting color using RGB values and line thickness
        t.pen(pencolor=(r,g,b), pensize=10)
        
        random.choice(moves)()
    
#to make the turtle move only after you click on the screen
#onscreenclick requires (x,y) parameters even if you do not use them.
#that's why function move_turtle has (x,y) parameters.

screen.onscreenclick(move_turtle)

#this will keep the window open even after it is complete
screen.mainloop()
```

Initially, this is what I wrote:

`random_moves = [t.forward(distance), t.backward(distance), t.right(angle), t.left(angle)]`

Idea was to use `random.choice[moves]` and execute a movement.

But this failed. Apparently, whenever Python sees`()` it executes the function immediately. So rather than generating a list of
movements, a list of `None` was created. Because after the movement is executed, it returns `None`.

So you have to create new functions with the movement methods and use them in the list.

Hence in the line, using pre-defined functions, a list was created: `moves = [move_forward, move_back, turn_right, turn_left]`. 

Theres is no `()` after each function. This means that the functions are only stored, but not executed.

In this is line : `random.choice(moves)(), (moves)` is followed by `()` to actually execute the function.

### More on Higher Order Functions

```python
def add(n1,n2):
    return n1 + n2

def subtract(n1,n2):
    return n1 - n2

def multiply(n1,n2):
    return (n1 * n2)

def divide(n1,n2):
    return (n1/n2)
```

```python
#higher order function

def calculate(n1,n2,operation):
    return operation(n1,n2)
```

```python
result = calculate(2,3,add)
result
```

```python
result = calculate(2,3,multiply)
result
```

Here,firstly, you have functions defined some basic math operations.

Then, there is another function `calculate` that takes the basic operation functions as an input.
This is known as a **Higher Order Function**.

It is not availbale in some other programs, but in Python, it is quite common.
