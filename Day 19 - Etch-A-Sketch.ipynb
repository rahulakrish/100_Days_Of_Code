## Etch-A-Sketch

Control the movement of the `turtle` object by `"listening"` for events like key-strokes, mouse-clicks etc.

You will use `screen event methods` for this.

```python
%%writefile etch-a-sketch.py

from turtle import Turtle, Screen
import random

t = Turtle()
screen = Screen()

#defining movements
def move_forward():
    t.forward(50)
    
def move_backward():
    t.backward(50)
    
def turn_right():
    t.right(90)
    
def turn_left():
    t.left(90)
    
def reset_screen():
    screen.resetscreen()
    
#setting up the objec to listen for event      
screen.listen()


#setting the keys for the object to move
screen.onkey(fun=move_forward,key="w")
screen.onkey(fun=move_backward,key="s")

screen.onkey(fun=turn_right,key="d")
screen.onkey(fun=turn_left,key="a")

#reset will clear screen with whatever drawing and reset the position of the turtle object to it's starting position
screen.onkey(fun=reset_screen,key="r")

screen.mainloop()    
```
