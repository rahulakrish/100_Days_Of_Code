Turtle Race

```python
%%writefile turtle_race.py

from turtle import Turtle,Screen
import random

#wrapping the whole thing inside a function to import as a module.
# this is the only way you can get input() to work. if you use regular !python,
# it will not work.

def main():
    #asking the user to bet on which turtle will win
    user_bet = input('Place your bet! Which turtle do you thing will win?!\n \
    Choose from the following: Red, Orange, Green, Black, Purple\n').lower()

    screen = Screen()

    #defining the screen size and setting limits
    screen.setup(width=600,height=400)

    #colors of the participating turtles
    colors = ['red','orange', 'black', 'green', 'blue', 'purple']
    
    #creating a list to iterate thru for the race
    turtle_list = []

    #initializing positions and colors for the first 3 turtles
    color_index = 0
    y_cord = 0

    #this loop for the turtle to get into position in the positive quadrant
    for tees in range(0,3):
        t = Turtle(shape='turtle')
        t.color(colors[color_index])
        color_index += 1
        t.penup()
        #this is the position of the first turtle
        t.goto(x=-275,y=y_cord)        
        #subsequent turtles will move to the same x position but higher
        y_cord += 75
        #appending to the list which will be used come race time
        turtle_list.append(t)
    
    #initializing positions and colors for the remaining turtles
    color_index = 3
    #y cordinate is -ve i.e. below the first turtle
    y_cord = -75 

    for tees in range(0,2):    
        t = Turtle(shape='turtle')
        t.color(colors[color_index])
        color_index += 1
        t.penup()
        t.goto(x=-275,y=y_cord)
        #the second turtle is further down 
        y_cord -= 75
        #appending to the list which will be used come race time
        turtle_list.append(t)

 
    #list of steps to move the turtle forward
    steps = [5,10,20]

    #flag for the while loop
    is_end = False

    while is_end == False:
        #pick a random turtle from the list of turtles
        for chosen_turtle in turtle_list:
            chosen_turtle = random.choice(turtle_list)
            #that turtle will move a random step forward
            chosen_turtle.forward(random.choice(steps))
            #in this logic, some turtles might not get chosen at all and some turtles can get chosen twice.
            #if you want to be fair, then remove the random.choice(turtle_list). 

            
            #setting the finish line as 200
            if chosen_turtle.xcor() >= 200:
                #changing the flag and ending the loop
                is_end = True
                print('\n')
                #getting the color of the turtle to declare the winner. title() will capitalize the first letter.
                winner = chosen_turtle.pencolor().title()
                print(f'{winner} color turtle is the winner!')
                break



    #check with the user's bet
    if chosen_turtle.pencolor() == user_bet:
        print('\n')
        print('Great call!You win!')
    else:
        print('\n')
        print(f'You lose! {winner} color turtle won!')



    screen.mainloop()
    
if __name__ == '__main__':
    main()
```
