## Beautiful Soup

- is used to parse data from `HTML` and `XML` files
- this way, you can pull specific tags, text, links etc. for further use

```python
from bs4 import BeautifulSoup
```

```python
#just using the read-write method
with open('./bs4-start/website.html') as file:
    contents = file.read()
print(contents)
```

```python
#using BeautifulSoup
soup = BeautifulSoup(contents, "html.parser")
soup
```

```python
#use prettify to get a nested structure
print(soup.prettify())
```

```python
#look up the title tag
print(soup.title)
```

```python
#look up h1 tag
print(soup.h1)
```

```python
#look up ul tag
print(soup.ul)
```

```python
#get all the text in the html file
print(soup.get_text())
```

```python
#get all the links in the HTML file
print(soup.a)
```

### find_all

The previous methods will only list the `first` instance of whatever you're asking to look up.

If you want `all` the instances of a particular tag, then use the `find_all` method

```python
soup.find_all("a")
```

```python
#get all the text in the anchor tags
for anchor_tag in soup.find_all('a'):
    print(anchor_tag.get_text())
```

```python
#get all the links in the anchor tag
for anchor_tag in soup.find_all('a'):
    print(anchor_tag.get('href'))
```
