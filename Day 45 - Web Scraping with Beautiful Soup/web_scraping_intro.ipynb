### Web Scraping Intro

- how to "scrape" information from a web page using `BeautifulSoup`
- this is the website: https://news.ycombinator.com/news
- find the most popular article at the moment and it's link

```python
from bs4 import BeautifulSoup
import requests
```

```python
response = requests.get('https://news.ycombinator.com/news')
print(response.text)
```

```python
#parse using BeautifulSoup
soup = BeautifulSoup(response.text,'html.parser')
print(soup.prettify())
```

Get all the titles from the page.
Upon inspection, titles have the following properties:
     `span` tag and `titleline` class

```python
#find_all will return a list of the elements having the span tag and the titleline class
titles = soup.find_all('span',class_='titleline')

#is a much bigger list, of course. checking the first 3 alone to get an idea
titles[0:3]
```

Now, loop thru this list to get all the anchor tags and the links to each article

```python
links = []
article_titles = []

for item in titles:
    anchor_tag = item.find('a') #anchor tag
    
    #get link and append to the list
    article_link = anchor_tag.get("href")
    links.append(article_link)
    
    #get the text and append to the list
    title_text = anchor_tag.get_text()
    article_titles.append(title_text)
```

```python
links[0:3]
```

```python
article_titles[0:3]
```

Next, check the number of votes each article has

```python
#article upvotes
article_upvotes = soup.find_all('span', class_='score')
article_upvotes[0:3]
```

```python
upvotes =  []

#loop to get the count
for upvote in article_upvotes:
        text = upvote.get_text()
        
        #get_text will return a string. split it to get the number
        split = text.split()
        
        #convert to int 
        votes = int(split[0])   
        
        #append to tthe list
        upvotes.append(votes)        
```

```python
upvotes[0:3]
```

Next, get the article with the most votes and it's link

```python
max_vote = max(upvotes)
max_vote_index = upvotes.index(max_vote)
max_vote_index
```

```python
print(f' Most popular article today is: {article_titles[max_vote_index]}')
print(f' It has received {upvotes[max_vote_index]} votes')
print(f' Here is the link to the article: {links[max_vote_index]}')
```
