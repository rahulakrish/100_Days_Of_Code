#### requests.put

```python
#update the graph
#PUT - /v1/users/<username>/graphs/<graphID>
#the values in the angle brackets are to be updated
url = f'https://pixe.la/v1/users/{USERNAME}/graphs/graph2'

params = {'date': '20260322',
         'quantity': '1'}

repsonse = requests.post(url=url, json=params, headers=headers )
```

```python
repsonse.text
```

```python
#writing a loop to update the past 4 days
start = '202603'
for _ in range(23,27):
    params = {'date': start + str(_),
             'quantity': '1'}
    repsonse = requests.post(url=url, json=params, headers=headers )   
```
