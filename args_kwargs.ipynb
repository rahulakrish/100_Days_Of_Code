### args & kwargs

`*args` - **many positional arguments**

If you see this in a function, that means it can take unlimited arguments

```python
def add(n1,n2):
    result = n1+ n2
    print(result)
    
add(2,3)
```

In the above function, you can provide only 2 numbers to add. What if you want to add 10? Or 50? Or 100? Obviously, you cannot type n1,n2,n3 etc.

What you can use instead, is `*args`. This is Python's way of specifying that the function can take **unlimited** number of arguments.

```python
def add(*args):
    print(type(args))
    sum = 0
    for n in args:
        sum += n
    return sum

add(2,4,5,6,7,8,9)
```

As you can see, you can input how many every arguments you want. `*args` returns a tuple that you can then loop theu and perform operations you want.

`**kwargs` - **many key word arguments

Just like how you an provide unlimited positional arguments, you can also do unlimited keyword arguments

```python
def calculate(n, **kwargs):
    print(kwargs)
    n += kwargs['add']
    n *= kwargs['multiply']
    return n
```

```python
calculate(5, add=5, multiply=5)
```

- `n` is a positional argument
- `**kwargs` returns a dictionary. Not a `tuple` like `*args`
- you're adding 5 to `n` and then multiplying the result by 5

This way you can specify unlimited keyword arguments.

You can also use this to build `Classes`:

```python
class Car():
    def __init__(self,**kwargs):
        self.make = kwargs['make']
        self.model = kwargs['model']
```

```python
t = Car(make='tesla',model='y')
print(t.make)
print(t.model)
```
