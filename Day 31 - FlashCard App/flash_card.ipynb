```python
from tkinter import *
import random
import pandas as pd
import csv
```

```python
df = pd.read_csv(r'C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\data\french_words.csv')
df.head()
```

```python
#constanst set-up
BACKGROUND_COLOR = "#B1DDC6"
frontimage_path = r"C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\images\card_front.png"
backimage_path = r"C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\images\card_back.png"
right_image_path = r"C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\images\right.png"
wrong_image_path = r"C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\images\wrong.png"
finish_button_path = r"C:\Users\e40050338\Documents\Python\Day 31 - FlashCard App\images\finish.png"
```

```python
# ----------------------------------------- UI Setup ------------------------------------------------- #
window = Tk()
window.title("Flash App")
window.config(padx=50, pady=50,bg = BACKGROUND_COLOR)

canvas = Canvas(width=800, height=526, highlightthickness=0,bg = BACKGROUND_COLOR )
frontimage = PhotoImage(file=frontimage_path)
backimage = PhotoImage(file=backimage_path)

canvas.grid(row=1,column=1,columnspan=3)

#creating global variables to be used later
current_word = None
french_image = None
french_heading = None
french_word = None
english_word = None


# ----------------------------------------- WORD setup ------------------------------------------------- #

#function to pick a random french word first
def pick_a_random_french_word():
    random_word_row_number = random.randint(0,len(df)-1)
    random_french_word = df.iat[random_word_row_number,0]
    
    return random_french_word

#function to display the french word in the canvas
def show_french_word():
    
    #statement to declare that you're modifying the global variables
    global current_word, french_image, french_heading, french_word
    
    #get a random french word
    current_word = pick_a_random_french_word()
    
    #create the image for the Frenchword    
    french_image = canvas.create_image(400,270, image=frontimage)
    
    #create the heading "French"
    french_heading = canvas.create_text(400,150,text="French",font=('Ariel',40,'italic'))
    #don't need the grid here apparenetly. the text just shows up in the right spot 
    
    #dipslay the french word - current_word
    french_word = canvas.create_text(400,263,text=current_word,font=('Ariel',60,'bold'))
    #don't need the grid here apparenetly. the text is in the right spot
    
    #change the states of the buttons
    #whenver the French word is displayed, you do not want the user to click on the buttons till he/she sees the english word
    #otherwise, this will mess up the export.
    wrong_button.config(state='disabled')
    right_button.config(state='disabled')
    
    #this will show the english word after 3 seconds
    window.after(3000, show_english_word)
    

def show_english_word():
    
    #modifying the english word
    global english_word 
    
    #getting the english word from the df
    english_word = df.loc[df["French"] == current_word, "English"].values[0]

    #modifying all the french items in the canvas to english
    canvas.itemconfig(french_image,image=backimage)
    canvas.itemconfig(french_heading,text='English')
    canvas.itemconfig(french_word,text=english_word)
    
    #change the states of the buttons
    #after the English word is displayed, then the buttons become active allowing the user to click on the appropriate one
    wrong_button.config(state='active')
    right_button.config(state='active')
    
# ----------------------------------------- BUTTON function ------------------------------------------------- #

#if the user clicks the right button, then a new french word is dispalyed
def right_button_function():
    show_french_word()
    
    
#if the user clicks the wrong button:
# 1. the current french word and the english translation is exported
# 2. a new french word is displayed
def wrong_button_function():  
    export_df = pd.DataFrame([[current_word,english_word]],columns=['French', 'English'])
    export_df.to_csv('words_to_learn.csv', mode='a', index=False, header=False)
    
    show_french_word()

                                                                        
# ----------------------------------------- BUTTON setup ------------------------------------------------- #

wrong_image = PhotoImage(file=wrong_image_path)
wrong_button = Button(image=wrong_image,command=wrong_button_function)
wrong_button.grid(row=3,column=2)

right_image = PhotoImage(file=right_image_path)
right_button = Button(image=right_image,command=right_button_function)
right_button.grid(row=3,column=3)

# ----------------------------------------- RUN ------------------------------------------------- #
show_french_word()

window.mainloop()

```
