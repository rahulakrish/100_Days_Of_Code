```python
"""

ARIA - Accessible Rich Internet Applications

It is a set of special attributes added to HTML tags to help people with disabilities navigate websites using assistive technologies, 
like screen readers (software that reads the screen out loud).

Why do we need ARIA? (The Problem)
Back in the day, HTML was simple: a <button> was a button, and an <a> was a link.

But modern websites are complex. Developers often use a generic <div> or <span> tag, style it with CSS to look like a beautiful toggle switch, and use JavaScript to make it clickable.

To a sighted user, it looks and acts like a switch. But to a screen reader, it’s just a generic, meaningless <div>. The screen reader user has no idea they can interact with it.
ARIA fixes this.

The Three Pillars of ARIA
ARIA attributes generally fall into three categories. When you are writing Selenium scripts, these are goldmines for locating elements.

1. Roles (role="...")
This tells the assistive tool what the element actually is, overriding its default HTML tag.

Example: <div role="button">

Meaning: "Hey screen reader, even though this is a div tag, treat it exactly like a clickable button."

2. Labels and Descriptions (aria-label, aria-labelledby)
This provides a text alternative for elements that don't have visible text on the screen (like icon-only buttons).

Example: <button aria-label="Close window">X</button>

Meaning: Instead of reading just the letter "X", the screen reader will say "Close window button".

💡 Selenium Tip: This is amazing for automation! Instead of trying to click a vague button tag, you can use:
driver.find_element(By.XPATH, "//button[@aria-label='Close window']")

3. States (aria-checked, aria-expanded, aria-disabled)
This tells the user the current condition of the element, which usually changes dynamically with JavaScript.

Example: <div role="checkbox" aria-checked="true">

Meaning: Tells the user that this custom checkbox is currently checked.

💡 Selenium Tip: You can use these states in Selenium to verify if a dropdown has successfully opened or if a checkbox was successfully clicked:
dropdown = driver.find_element(By.XPATH, "//div[@aria-expanded='true']")

The Golden Rule of ARIA
There is a famous saying among web developers: "No ARIA is better than bad ARIA." If a developer can use a native HTML element (like <button>), they should. 
They should only use ARIA when they are building custom, complex components (like tabs, sliders, or pop-up menus) that native HTML can't handle out of the box.

Because these attributes are meant to describe the page explicitly, they make fantastic, reliable hooks for your Selenium locators since they rarely change layout-wise!

"""

;
```
