Difference b/w `Window` and `Canvas`:

`Window`:

- The window is the main application container.

- It’s the top-level frame created by Tk()

- Has the title bar, close/minimize buttons, size, etc.

- Holds everything inside your app

`Canvas`:

- The canvas is a widget that lives inside the window.

- Used for drawing text, shapes, images, animations

- Very flexible for games, clocks, diagrams, custom UIs

- You place it inside the window

```python
from tkinter import *
import math

# ---------------------------- CONSTANTS ------------------------------- #
PINK = "#e2979c"
RED = "#e7305b"
GREEN = "#9bdeac"
YELLOW = "#f7f5dd"
FONT_NAME = "Courier"
WORK_MIN = 1
SHORT_BREAK_MIN = 5
LONG_BREAK_MIN = 20
reps = 0

# ---------------------------- TIMER RESET ------------------------------- # 

# ---------------------------- TIMER MECHANISM ------------------------------- # 

def start_timer():
    global reps
    reps += 1
    
    
    work_sec = WORK_MIN * 60
    short_break_sec = SHORT_BREAK_MIN * 60
    long_break_sec = LONG_BREAK_MIN * 60
    
    #if its 8th rep:
    if reps % 8 == 0:
        count_down(long_break_sec)
        canvas.itemconfig(headline,text='Break',fill=PINK)
        
    #if its 2nd/4th/6th rep:
    elif reps % 2 == 0:
        count_down(short_break_sec)
        canvas.itemconfig(headline,text='Break',fill=RED)
        
        
     #if its 1st/3rd/5th/7th rep:
    else:
        count_down(work_sec)
        canvas.itemconfig(headline,text='WORK',fill=GREEN)


# ---------------------------- COUNTDOWN MECHANISM ------------------------------- # 

def count_down(count):
    #to display in minutes:seconds format
    count_min = math.floor(count / 60) #this is the get the minutes
    count_sec = count % 60 #this is to get the seconds
    
    #this if condition is to get the seconds to display in the format "00"
    if count_sec < 10: #notice how you are changing the type from int to text here
        count_sec = f'0{count_sec}'
   
    #if you want to modify somthing in the canvas, then you need to use
    #the itemconfig method. By assigning the text to a variable called timer_text
    #you can modify it to display the timer.
    canvas.itemconfig(timer_text, text= f'{count_min}:{count_sec}')
    
    #this is to start the counter, here you subtract by 1 second.
    if count>0:
        window.after(1000, count_down, count - 1)
    else:
        start_timer()

    
# ---------------------------- UI SETUP ------------------------------- #

#create a new window
window = Tk()
#set the title
window.title('Pomodoro')
window.config(padx=100,pady=50,bg=YELLOW)

#A canvas is a rectangular area intended for drawing pictures or other complex layouts.
#On it you can place graphics, text, widgets, or frames.
#canvas is different from window. Do not confuse the two
canvas = Canvas(width=200, height=400, bg=YELLOW, highlightthickness=0)

#use the PhotoImage class of Tkinter to import the image
tomato_img = PhotoImage(file='tomato.png')

#paste the image on the canvas the numbers are for x,y position of the image
#the numbers are approx. half of the values of the canvas window to center the image
canvas.create_image(100,200,image=tomato_img)

#create the text of the timer
timer_text = canvas.create_text(100,220,text="00:00",fill='white', font=('Times',36))

#define position of timer text using grid
canvas.grid(row=0,column=1)

#TIMER headline
headline = canvas.create_text(100,20,text="Timer",fill=GREEN, font=('Times', 44))

#you could've also created the TIMER headline using the Label class.
#timer_label = Label(text='Timer')
#timer_label.grid(xx,xx)

check_mark = '✅'

#function for the stop button
#setting the global variable
increment = 0

def stop_button_clicked():
    global increment
    x_pos = 50 + increment
    canvas.create_text(x_pos,350,text=check_mark,fill='purple',tags='checkmarks')
    increment += 15
    canvas.grid(row=4,column=1)
    
#function for the reset button
def reset_button_clicked():
    canvas.delete('checkmarks')
    

#defining the start button 
start_button = Button(text='START', command=start_timer, bg='white', font=('Times',8,'bold'))
start_button.grid(row=5,column=0)

#defining the stop button
stop_button = Button(text='STOP', command=stop_button_clicked, bg='white', font=('Times',8,'bold'))
stop_button.grid(row=5, column=1)

#defining the reset button
reset_button = Button(text='RESET', command=reset_button_clicked, bg='white', font=('Times',8,'bold'))
reset_button.grid(row=5,column=2)


#mainloop checks every millisecond if somehting happened in the window
#In a GUI program, you need to check for "events" i.e. did the user click a button
#if the user entered something into a text box etc.
window.mainloop()
```
