```python
%%writefile paddle.py

#class to create Paddles
from turtle import Turtle

class Paddle(Turtle):
    
    #you have position as an argument.
    #this way you can specify where you want the paddles created i.e left paddle on the left of the screen and 
    #similarly for the right paddle
    
    def __init__(self, position):
        super().__init__()
        self.shape("square") 
        self.color("white")
        self.shapesize(stretch_wid=5, stretch_len=1) #making the object long
        self.penup()
        self.goto(position)#telling the object where to go

    #controlling up and down movements of the paddle object    
    def go_up(self):
        new_y = self.ycor() + 20 #moving it by 20 pixel increments
        self.goto(self.xcor(), new_y)

    def go_down(self):
        new_y = self.ycor() - 20
        self.goto(self.xcor(), new_y)
```

```python
%%writefile ball.py

#class to build the Ball
from turtle import Turtle

class Ball(Turtle):

    def __init__(self):
        super().__init__()
        self.color("white")
        self.shape("circle")
        self.penup()
        self.x_move = 3 #move increments in x
        self.y_move = 3 #move increments in y
        self.move_speed = 1 #speed

    #moving the ball     
    def move(self):
        new_x = self.xcor() + self.x_move #telling it how much to move in x
        new_y = self.ycor() + self.y_move #telling it how much to move in y
        self.goto(new_x, new_y) 
        
    #this is to make it bounce off walls. When it hits a wall, the ball will move in the opposite y direction 
    def bounce_y(self):
        self.y_move *= -1

    #this is to make it bounce off paddles. When it hits a paddle, the ball will move in the opposite x direction            
    def bounce_x(self):
        self.x_move *= -1
        #this is to make the ball move slightly faster after it hits the paddle. makes the game more exciting
        self.move_speed *= 0.5
        
    #this is to reset the game when the ball misses either paddle
    def reset_position(self):  
        self.goto(0, 0)
        #this is to reset to the initial speed
        self.move_speed = 0.1
        #make it move in the opposite direction i.e if the right misses, then the ball resets and moves left
        self.bounce_x()
```

```python
%%writefile scoreboard.py

from turtle import Turtle


class Scoreboard(Turtle):

    def __init__(self):
        super().__init__()
        self.color("white")
        self.penup()
        self.hideturtle()
        self.l_score = 0
        self.r_score = 0
        self.update_scoreboard()

    def update_scoreboard(self):
        self.clear()
        self.goto(-100, 200)
        self.write(self.l_score, align="center", font=("Courier", 80, "normal"))
        self.goto(100, 200)
        self.write(self.r_score, align="center", font=("Courier", 80, "normal"))

    #update score    
    def l_point(self):
        self.l_score += 1
        self.update_scoreboard()

    def r_point(self):
        self.r_score += 1
        self.update_scoreboard()
```

```python
%%writefile main.py

from turtle import Screen, Turtle
from paddle import Paddle
from ball import Ball
from scoreboard import Scoreboard
import time

screen = Screen()
screen.bgcolor("black")
screen.setup(width=800, height=600)
screen.title("Pong")
screen.tracer(0)

r_paddle = Paddle((350, 0))
l_paddle = Paddle((-350, 0))
ball = Ball()
scoreboard = Scoreboard()

screen.listen()
screen.onkey(r_paddle.go_up, "Up")
screen.onkey(r_paddle.go_down, "Down")
screen.onkey(l_paddle.go_up, "w")
screen.onkey(l_paddle.go_down, "s")

game_is_on = True
while game_is_on:
    time.sleep(0.03)
    screen.update()
    ball.move()

    #Detect collision with wall
    if ball.ycor() > 280 or ball.ycor() < -280:
        ball.bounce_y()

    #Detect collision with paddle
    if ball.distance(r_paddle) < 50 and ball.xcor() > 320 or ball.distance(l_paddle) < 50 and ball.xcor() < -320:
        ball.bounce_x()

    #Detect R paddle misses
    if ball.xcor() > 380:
        ball.reset_position()
        scoreboard.l_point()

    #Detect L paddle misses:
    if ball.xcor() < -380:
        ball.reset_position()
        scoreboard.r_point()

screen.exitonclick()
```
