```python
%%writefile main_turtle.py

from turtle import Turtle

class MainTurtle(Turtle):
    
    def __init__(self):
        super().__init__()

        self.shape("turtle")
        self.penup()
        self.goto(x = 0, y = -280)
        self.setheading(90)
        
    def move_up(self):
        self.forward(distance = 10)
        if self.ycor() >= 280:
            self.goto(x = 0, y = -280)
        
    def move_down(self):
        self.backward(distance = 10)
        
```

```python
%%writefile car_manager.py

from turtle import Turtle
import random

COLORS = ['red','green','yellow','blue']
MOVING_DISTANCE = 10

class CarManager(Turtle):
    
    def __init__(self):
        super().__init__()
        
        self.all_cars = []
    
    def create_car(self):
        
        x_position = 360
        y_position = random.randint(-350,350)
        new_car = Turtle()
        new_car.shape('square')
        new_car.color(random.choice(COLORS))
        new_car.shapesize(stretch_len=2)
        new_car.penup()
        new_car.goto(x_position,y_position)
        self.all_cars.append(new_car)
        
    def move_car(self):
        for car in self.all_cars:
            car.backward(MOVING_DISTANCE)
            
        
```

```python
%%writefile main.py

from turtle import Screen
from main_turtle import MainTurtle
from car_manager import CarManager
import time
screen = Screen()
screen.setup(width=800,height=600)
screen.tracer(0)

main_turtle = MainTurtle()
car_manager = CarManager()

screen.listen()
screen.onkey(main_turtle.move_up, "Up")
screen.onkey(main_turtle.move_down, "Down")

last_car_creation_time = time.time()
car_creation_delay = 0.5

game_on = True
while game_on:
    time.sleep(0.05)
    screen.update()
    
    current_time = time.time()
    
    if current_time - last_car_creation_time >= car_creation_delay: 
        car_manager.create_car()
        last_car_creation_time = current_time
    
    car_manager.move_car()

screen.exitonclick()
```
