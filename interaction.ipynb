Here, you will look at how to accomplish tasks in the browser\
For eg: click on a link, input text in the search bar etc.; \
all things you would normally do on a website.

```python
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
```

```python
#using the xpath to get the article_count from the wikipedia page
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

article_count = driver.find_element(By.XPATH, '/html/body/div[3]/div/div[3]/main/div[3]/div[3]/div[2]/div[1]/div/div[3]/ul/li[2]/a[1]')

print(article_count.text)

driver.quit()
```

```python
#another way but does not help since the articlecount id and more than 1 item in it
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

article_count = driver.find_element(By.ID, 'articlecount')

print(article_count.text)

driver.quit()

#here, you will have to modify the string to get your output
```

Note about CSS_SELECTOR:\
When trying to find a class using CSS_SELECTOR and you have an ID then, this is how it will be:\
`driver.find_element(By.CSS_SELECTOR, #ID_name)`\
\
For an ID, you have to use #. The above line is equivalent to:\
`driver.find_element(By.ID, #ID_name)`\
\
Similarly, for class:\
`driver.find_element(By.CSS_SELECTOR, .class_name)`\
\
The # is replaced by .\
This is equivalent to:\
`driver.find_element(By.class_name, .class_name)`

```python
#a working example
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

article_count = driver.find_elements(By.CSS_SELECTOR, '#articlecount a')

for x in article_count:
    print(x.text)

driver.quit()

#Since you're getting a list, use list indexing to get the article count
```

How to click on a link using selenium

```python
#how to click on a link using selenium
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

article_count = driver.find_elements(By.CSS_SELECTOR, '#articlecount a')

#from the above you know the number of ariticles is the 2nd item
#use the click method to click on the link
article_count[1].click() 

driver.quit()
```

Since clicking on a link is a very common task, Selenium has a method called `link_text` to make this easy.\
All you have to do is type in the text of the link, and Selenium will look for it on the webpage and click.\
That's it. \
No need to try and find the element in the HTML code.

```python
#a working example
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

#use LINK_TEXT
article_count = driver.find_element(By.LINK_TEXT,'7,183,795')

article_count.click()
```

How to type using Selenium

```python
#For eg: typing Python into the search bar on Wikipedia. The search bar elment has the name attribute "search"
driver = webdriver.Firefox()

driver.get('https://en.wikipedia.org/wiki/Main_Page')

search_box = driver.find_element(By.NAME,'search')

#use SEND_KEYS method to type into the search box
search_box.send_keys('Python')

#hit ENTER after typing into the search box
# if you want to send a key that's not a letter, number of symbol,
#you need to import the keys package
search_box.send_keys(Keys.ENTER)

```
