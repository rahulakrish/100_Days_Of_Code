### Basic OOP

```python
# Basic OOP

class BankAccount:
    def __init__(self,name,opening_balance):
        self.name = name
        self.balance = opening_balance
        
    def deposit(self,amount):
        self.amount = amount
        self.balance += amount
        
        
r = BankAccount("r",1000)
```

```python
print(r.name)
print(r.balance)
```

```python
r.deposit(1000)
print(r.balance)
```

### Class Inheritance

```python
class Animal:
    def __init__(self):
        self.num_eyes = 2
        
    def breathe(self):
        print("inhale,exhale")
        
class Fish(Animal): #for the initializer, you specify the class you want to inherit from.
    def __init__(self):
        super().__init__() # this line code is used to "inherit" all the attributes and methods from the Animal class
    
    #you can also still define other methods to the subclass. It will have everything it inherited + it's own methods.
    def swim(self):
        print('moving underwater')
```

```python
#creating an instance
nemo = Fish()

#methods 
nemo.swim() # the Fish classes' own method

nemo.breathe()# the method inherited from the Animal class

#attributes
nemo.num_eyes
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

#even though you are inherting methods and attributes from the parent class, in the child class, you still have to specify
# the same initializers. 
#That's because attributes exist only after __init__ runs + the child class does not run the parent class automatically.


class Student(Person):#setting it up to inherit from the Parent class.
    def __init__(self,name,age): #here, for the child class you are defining that is should have a name and age
        super().__init__(name,age)# and then, over here, you are telling it to get the name from the parent class.
        #super() refers to the parent class Person & the name,age in the brackets are to inherit the values from the Parent class. 
        
    def info(self,subject,grade):
        self.subject = subject
        self.grade = grade
```

```python
#Just using the Parent class, Person
x = Person("x",12)
print(x.name)
print(x.age)
```

```python
#using the child class, Student
x = Student("x",12)
```

```python
print(x.name)
print(x.age)

x.info('math',85)
print(x.subject)
print(x.grade)
```

```python
class Car():
    
    def __init__(self,name,drivetrain):
        self.name = name
        self.drivetrain = drivetrain
        
    def start_car(self):
        print(f'{self.name} has started')


class NewCar(Car):
    
    def __init__(self,name,drivetrain,model_year):#again, for the child class, you are specifying the attributes it should have.
        #you can also add your own attribute. 'model_year' is not an attribute from the parent class. Only 'name' & 'drivetrain'
        #are
        super().__init__(name,drivetrain)
        self.model_year = model_year
              
    def stop_car(self):
            print(f'{self.name} has stopped')       
```

```python
#using the parent class
t = Car('tesla', 'EV')
print(t.name)
t.start_car()
```

```python
a = NewCar('honda','IC','2025')
print(a.name)
print(a.drivetrain)
print(a.model_year)
a.stop_car()
```
