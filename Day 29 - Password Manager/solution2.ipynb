```python
from tkinter import *
from tkinter import messagebox
import string
import random
import os
path = "C:\\Users\\e40050338\\Documents\\Python\\Day 29 - Password Manager\\data_file_2.txt"
```

```python

# ---------------------------- PASSWORD GENERATOR ------------------------------- #


#function to generate the password
def generate_password():

    #get all the letters, digits and characters using string in-built method
    lowercase = string.ascii_lowercase
    uppercase = string.ascii_uppercase
    digits = string.digits
    punctuation = string.punctuation

    #use list comprehensions to pick 2 lowercase, 3 uppercase, 2 digits etc.
    lower_list = [random.choice(lowercase)for _ in range(0,2)]
    upper_list = [random.choice(uppercase)for _ in range(0,3)]
    digit_list = [random.choice(digits)for _ in range(0,2)]
    punct_list = [random.choice(punctuation)for _ in range(0,4)]

    #join everthing together
    cumulative = lower_list + upper_list + digit_list + punct_list
    
    #shuffle the order
    random.shuffle(cumulative)

    #join them into one string 
    word = " ".join(cumulative)
    
    #remove the spaces in-between
    password = word.replace(" ","")
    
    return password


#function to update the password entry
#here you will run the generate_password function and use the output to update the entry.
#therefore in the Button widget, you will invoke this function and not the generate_password function.
def update_password_entry():
    #run the generate_password function
    password_value = generate_password()
    
    #delete any existing entry
    #syn: delete(start_index,end_index)
    password_entry.delete(0, END)
    
    #insert the generated pwd into the entry
    password_entry.insert(0,password_value)
    

# ---------------------------- RETRIEVE INPUTS FROM THE ENTRY WIDGETS ------------------------------- #
#function to fet inputs from all the entries    
def retrieve_inputs():
    website_name = website_entry.get()
    
    password_generated = password_entry.get() 
   
    email_username = email_entry.get()

    return website_name, email_username, password_generated

# ---------------------------- SAVE PASSWORD ------------------------------- #

#run the retrieve_inputs function here to get each value
def save_password():
    
    #get the outputs from the retrieve_inputs function
    website_name, email_username, password_generated = retrieve_inputs()
    
    
    #adding Validation: no field should be empty. verfy using len of the string
    if len(website_name) != 0 and len(password_generated) != 0:
        
        #like VBA, tkinter has a message box function that returns a boolean True/False for Yes/No button clicks
        confirmation =  messagebox.askokcancel(title="Confirm to proceed",
                                               message=f"Website Name: {website_name}\n"
                                                       f"Email Username: {email_username}\n"
                                                       f"Password: {password_generated}"
                                                )
        
        #if True
        if confirmation:                          
            if os.path.exists(path):
                with open("data_file_2.txt", mode = 'a') as file:
                    file.write(f'\n{website_name} | {email_username} | {password_generated}')
            else:
                 with open("data_file_2.txt", mode = 'w') as file:
                    file.write(f'\n{website_name} | {email_username} | {password_generated}')
    
    #validation for empty fields
    else:     
        messagebox.showinfo(title="Empty field", message='Please do not leave any fields empty!')
        
        

# ---------------------------- UI SETUP ------------------------------- #
#create a window instance
window = Tk()

#set the title of the window
window.title("Password Manager")

#add some spacing in the x and y direction b/w the canvas and the window
window.configure(padx=20,pady=20,bg='white')

#create a canvas
#highlightthickness is used to blur out the edges of the image
canvas = Canvas(width=200,height=200,bg="white",highlightthickness=0)

#the following steps are used to place an image onto the canvas
#create a image object of the logo using PhotoImage class
myimage = PhotoImage(file='logo.png')
#bring it onto the cnavas 
canvas.create_image(100,100,image=myimage)
canvas.grid(column=1,row=0)


#website label
website_label = Label(text="Website:",bg="white",font=('Courier',14))
website_label.grid(column=0,row=1,sticky='e')
#The sticky argument in the Tkinter grid() method is used to control the alignment and resizing behavior of a widget within its allocated cell(s).
#By default, if a cell is larger than the widget it contains, the widget is centered within the cell. 
#e-East - sticks to the right edge, w- West, n-North, s-South

#email\username label
email_label = Label(text="Email\\Username:",bg="white",font=('Courier',14))
email_label.grid(column=0,row=2)

#password label
password_label = Label(text="Password:",bg="white",font=('Courier',14))
password_label.grid(column=0,row=3,sticky='e')

#generate password button - invoke the update_password_entry function here
generate_button = Button(text="Generate Password",font=('Courier',13),command = update_password_entry)
generate_button.grid(column=2,row=3)

#add button - invoke the save_password function here
add_button = Button(text="Add",font=('Courier',13),width=36,command = save_password)
add_button.grid(column=1,row=6,columnspan=2)

#website_entry
website_entry = Entry(width=35,bg='white')
website_entry.grid(row=1,column=1,columnspan=2,sticky='w')
website_entry.focus_set()

#email_entry
email_entry = Entry(width=35,bg='white',font=('Courier',11))
email_entry.insert(0,'dummy@email.com')
email_entry.grid(row=2,column=1,columnspan=2,sticky='w')

#password_entry
password_entry = Entry(width=21,bg='white')
password_entry.grid(row=3,column=1,columnspan=2,sticky='w')

window.mainloop()

```
