```python
%%writefile turtle_race.py

from turtle import Turtle,Screen

screen = Screen()
screen.setup(width=600,height=400)
colors = ['red','orange', 'black', 'green', 'blue', 'purple']


def start_race(x,y):
    turtle_list = []
    
    color_index = 0
    y_cord = 0

    for tees in range(0,3):
        t = Turtle(shape='turtle')
        t.color(colors[color_index])
        color_index += 1
        t.penup()
        t.goto(x=-275,y=y_cord)
        y_cord += 75
        turtle_list.append(t)
        
        
        


    color_index = 3
    y_cord = -75 

    for tees in range(0,2):    
        t = Turtle(shape='turtle')
        t.color(colors[color_index])
        color_index += 1
        t.penup()
        t.goto(x=-275,y=y_cord)
        y_cord -= 75
        turtle_list.append(t)
        
    


screen.onscreenclick(start_race)   
screen.mainloop()
```

```python
%%writefile race.py

from turtle import Turtle,Screen
import random

screen = Screen()
screen.setup(width=600,height=400)
colors = ['red','orange', 'black', 'green', 'blue', 'purple']

turtle_list = []

color_index = 0
y_cord = 0

for tees in range(0,3):
    t = Turtle(shape='turtle')
    t.color(colors[color_index])
    color_index += 1
    t.penup()
    t.goto(x=-275,y=y_cord)
    y_cord += 75
    turtle_list.append(t)

color_index = 3
y_cord = -75 

for tees in range(0,2):    
    t = Turtle(shape='turtle')
    t.color(colors[color_index])
    color_index += 1
    t.penup()
    t.goto(x=-275,y=y_cord)
    y_cord -= 75
    turtle_list.append(t)
    
# print(turtle_list)

steps = [5,10,15]
x_cor = 200

is_end = False

while is_end == False:

    for random_turtle in turtle_list:
        chosen_turtle = random.choice(turtle_list)
        chosen_turtle.forward(random.choice(steps))

        if chosen_turtle.xcor() == 200:
            is_end = True
   
    

screen.mainloop()
```
