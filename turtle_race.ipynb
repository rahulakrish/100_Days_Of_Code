## Turtle Race

```python
%%writefile turtle_race.py

from turtle import Turtle,Screen

screen = Screen()
screen.setup(width=600,height=400)
colors = ['red','orange', 'yellow', 'green', 'blue', 'purple']

color_index = 0
y_cord = 0

for tees in range(0,3):
    t = Turtle()
    t.shape('turtle')
    t.color(colors[color_index])
    color_index += 1
    t.penup()
    t.goto(x=-275,y=y_cord)
    y_cord += 75

    
color_index = 3
y_cord = -75 

for tees in range(0,2):    
    t = Turtle()
    t.shape('turtle')
    t.color(colors[color_index])
    color_index += 1
    t.penup()
    t.goto(x=-275,y=y_cord)
    y_cord -= 75

    
screen.mainloop()
```
