```python
#start the game
print('Welcome to the Higher vs Lower game! Here, you will guess who has the most Instagram followers (in millions)!')

print('\n')

#pull data from the dictionary using random indexes
import random

number1 = random.randint(0,49)

number2 = random.randint(0,49)

#this is just in case both numbers are equal. you run the loop again to generate a new number.
while number1 == number2:

    number2 = random.randint(0,49)

#use the random numbers as indexes to print the statements

print(f"Compare A: {data[number1]['name']}, a {data[number1]['description']}, from {data[number1]['country']}\n")

print('against\n')

print(f"        B: {data[number2]['name']}, a {data[number2]['description']}, from {data[number2]['country']}\n")

print('\n')
```

```python
#get user input
user_choice = input('Type A or B\n').lower()

#a function to define the first and second for comparision
def first_second(user_choice):

    if user_choice == 'a':
        first = data[number1]
        second = data[number2]
    elif user_choice == 'b':
        first = data[number2]
        second = data[number1]
        
    return first, second

#remember, it not enough to just define the function. you need to assign the outputs to a varaible to be able to use the results.
#return does that. It allows you the get the output from the function and use it outside the function.
#that's what first,second is for here.


#you're getting the data from the dictionary and using it for comparision
first,second = first_second(user_choice)


#keeping track of the score
score = 0

#flag for the while loop
game_over = False


#loop
while game_over == False:

    if first['follower_count'] > second['follower_count']:
        
        #this condition is when you need to run consecutive loops. If the 'b' is right option, then 'b' becomes the new 'a' in the 
        # next loop. 'b' which was set as number2, now becomes number1 in the next loop
        if user_choice == 'b':
            
            number1 = number2
        
        print(f"Correct! {first['name']} has {first['follower_count']} million followers! {second['name']} has only {second['follower_count']} million followers!")
        
       
        #add to the score
        score += 1
        
        #generate a new number for the next comparision
        number2 = random.randint(0,49) 
        
        #again, check to make sure both numbers are not the same
        while number1 == number2:
            
            number2 = random.randint(0,49) 
        
        #get data from the dictionary for the new comparision
        second = data[number2]
        
        print('\n')
        
        print(f"Compare A: {first['name']}, a {first['description']}, from {first['country']}\n")
        
        print('against\n')
        
        print(f"        B: {second['name']}, a {second['description']}, from {second['country']}\n")
        
        

        user_choice = input('Type A or B\n').lower()
        
        print('\n')

        first,second = first_second(user_choice)
        
    else:
        
        print("Sorry! You lose!")
        
        print('\n')
        
        print(f"{second['name']} has {second['follower_count']} million followers. {first['name']} has only {first['follower_count']} million followers! ")
        
        print('\n')
        
        print(f"Your score is {score}")
        
        game_over = True
```
