# Guessing Game - with attempt tracking

```
import random

#generate numbers between 1 and 100
number_list = list(range(1,101))

#pick a number from the list
chosen_number = random.choice(number_list)

# print(chosen_number)

# print('\n')

print("I've picked a number between 1 and 100. Can you guess the number?\n" )

#set difficulty level
difficulty_level = input('Choose difficulty level: easy or hard\n').lower()

if difficulty_level == 'easy':
    number_attempts = 10
else:
    number_attempts = 5

print ('\n')

print(f'You have {number_attempts} attempts\n')


#check for duplicates
guessed_list = []

#set flag for while loop
game_over = False

while game_over == False :

    #if user runs out of attempts
    if number_attempts == 0:

        print('Sorry, you lose!\n')

        break

    #start guessing
    user_guess = int(input('Guess a number\n'))

    #check for duplicates
    if user_guess not in guessed_list:

        guessed_list.append(user_guess)

    else:

        print('You already guessed this number.Try again!\n')


        number_attempts = number_attempts - 1

        print(f'You have {number_attempts} attempts left \n')

        continue

    if user_guess > chosen_number:

        #penalize the wrong attempt
        number_attempts = number_attempts - 1

        print(f'Too high. You have {number_attempts} attempts left \n ')

    elif user_guess < chosen_number:

        #penalize the wrong attempt
        number_attempts = number_attempts - 1

        print(f'Too low. You have {number_attempts} attempts left \n')


    if user_guess == chosen_number:

        game_over = True

        print('Awesome! You win')

```
